```python
df.columns
```

```python
df[['KHARIF SORGHUM PRODUCTION (1000 tons)', 'RABI SORGHUM PRODUCTION (1000 tons)', 'SORGHUM PRODUCTION (1000 tons)']].head()
```

```python
millet_cols=['SORGHUM PRODUCTION (1000 tons)',
'PEARL MILLET PRODUCTION (1000 tons)',
'FINGER MILLET PRODUCTION (1000 tons)']

# Total millet production column
df['Total millet production'] = df[millet_cols].sum(axis=1)

# Group by year and total millet production
millet_by_year = df.groupby('Year')['Total millet production'].sum().reset_index().sort_values(by='Year')

# Filter last 50 years
millet_last_50 = millet_by_year.tail(50)

# Find peak year to annotate
peak_row = millet_last_50.loc[millet_last_50['Total millet production'].idxmax()]

fig = px.line(
    millet_last_50,
    x='Year',
    y='Total millet production',
    title="Millet Production in India (Last 50 Years)",
    markers=True
)

fig.update_traces(
    line=dict(color='darkorange', width=2),
    marker=dict(color='darkorange', size=5),
    hovertemplate='<b>Year:</b> %{x}<br><b>Production:</b> %{y:,.0f} (1000 tons)<extra></extra>'
)

# Annotate only the peak year
fig.add_annotation(
    x=peak_row['Year'],
    y=peak_row['Total millet production'],
    text=f"Peak: {int(peak_row['Total millet production']):,}",
    showarrow=True,
    arrowhead=2,
    ax=40,
    ay=-40,
    font=dict(size=11, color='darkred'),
    bgcolor='lightyellow',
    bordercolor='darkorange'
)

fig.update_layout(
    xaxis_title='Year',
    yaxis_title='Millet Production (1000 tons)',
    hovermode='x unified'
)

fig.show()
```

## Insights: Millet Production in India (Last 50 Years)

- **No consistent upward trend** — unlike rice and wheat, millet production has fluctuated significantly, indicating vulnerability to climate shocks, policy shifts, and market forces.
- **Peak production occurred in the early 1970s–80s**, after which a long-term decline set in — driven by the Green Revolution's focus on wheat and rice, which reduced millet cultivation incentives.
- **Sharp dips** visible in drought years (e.g., early 1970s, mid-1980s, 2002) confirm that millets, despite being drought-tolerant, are still affected by extreme climate variability.
- **Gradual recovery post-2015** suggests renewed government interest in promoting millets as a nutrition-rich, climate-resilient crop (India declared 2023 the International Year of Millets).
- **Implication**: Millets are underutilized yet highly strategic — policy support, MSP incentives, and consumer awareness programs can help reverse the declining trend and strengthen food security.

Impact of Area Cultivated on Production (Rice, Wheat, Maize)

```python
crops = {
    'Rice':  ('RICE AREA (1000 ha)',  'RICE PRODUCTION (1000 tons)'),
    'Wheat': ('WHEAT AREA (1000 ha)', 'WHEAT PRODUCTION (1000 tons)'),
    'Maize': ('MAIZE AREA (1000 ha)', 'MAIZE PRODUCTION (1000 tons)')
}

yearly = df.groupby('Year')[
    [col for pair in crops.values() for col in pair]
].sum().reset_index()

# For each crop: pick its two columns, rename to Area/Production, tag with crop name
combined = pd.concat([
    yearly[[area_col, prod_col]]
        .rename(columns={area_col: 'Area', prod_col: 'Production'})
        .assign(Crop=crop)
    for crop, (area_col, prod_col) in crops.items()
], ignore_index=True)

fig = px.scatter(
    combined,
    x='Area',
    y='Production',
    color='Crop',
    trendline='ols',
    title='Impact of Area Cultivated on Production (Rice, Wheat, Maize)',
    labels={'Area': 'Area Cultivated (1000 ha)', 'Production': 'Production (1000 tons)'},
    hover_data=['Crop']
)

fig.update_layout(
    xaxis_title='Area Cultivated (1000 ha)',
    yaxis_title='Production (1000 tons)',
    legend_title='Crop',
    hovermode='closest'
)

fig.show()
```

The scatter plot shows how production changes as cultivated area increases for Rice, Wheat, and Maize. The **trendlines (OLS regression)** reveal the strength and slope of the relationship for each crop — a steeper line means higher yield efficiency per hectare.


## Insights: Impact of Area Cultivated on Production (Rice, Wheat, Maize)

- **All three crops show a strong positive correlation** — as cultivated area increases, production rises consistently, confirming that land expansion has been a key driver of output growth in India.
- **Rice has the steepest trendline**, indicating the highest production gain per additional hectare — driven by intensive irrigation and high-yield varieties (HYV) introduced during the Green Revolution.
- **Wheat closely follows Rice** in slope, reflecting similar HYV adoption in Indo-Gangetic plains; however, its points are more tightly clustered, showing consistent yields across years.
- **Maize has the flattest trendline** and most scattered points — suggesting higher yield variability, likely due to its use in both food and feed markets and varying farming intensities across states.
- **Implication**: Expanding cultivated area alone is not sufficient — improving yield per hectare through better seeds, irrigation, and tech is critical, especially for Maize which shows the most untapped potential.


Rice vs. Wheat Yield Across States

```python
data_yield = df.groupby("State Name")[[
    "RICE PRODUCTION (1000 tons)", "RICE AREA (1000 ha)",
    "WHEAT PRODUCTION (1000 tons)", "WHEAT AREA (1000 ha)"
]].sum().reset_index()


data_yield["Rice Yield"] = data_yield["RICE PRODUCTION (1000 tons)"] / data_yield["RICE AREA (1000 ha)"]
data_yield["Wheat Yield"] = data_yield["WHEAT PRODUCTION (1000 tons)"] / data_yield["WHEAT AREA (1000 ha)"]

data_yield = data_yield[["State Name", "Rice Yield", "Wheat Yield"]]
data_yield
```

```python
plt.figure(figsize=(12, 6))

# Melt the dataframe to long format for easy plotting
data_melted = data_yield.melt(id_vars="State Name", #values need to be fixed
                              value_vars=["Rice Yield", "Wheat Yield"], #cols in rows
                              var_name="Crop", 
                              value_name="Yield")

# Create  grouped bar chart
fig = px.bar(data_melted, 
             x="State Name", 
             y="Yield", 
             color="Crop", 
             barmode="group",
             title="Rice vs Wheat Yield Across States",
             labels={"Yield": "Yield (tons per hectare)"})

# Show chart
fig.show()
```
